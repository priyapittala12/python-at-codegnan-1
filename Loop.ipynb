```python
# check whether the number is divisible by 3 and 5.
lst = list(map(int, input().split()))
for n in lst:
    if n%3==0 and n%5==0:
        print("FIZZBUZZ")
    elif n%3==0:
        print("FIZZ")
    elif n%5==0:
        print("Buzz")
else:
    print(f'{n}')
        
```

```python
# break
lst = [11,12,9,5,6,7,13]
target = 5
n = len(lst)
for i in range(n):
    if lst[i]==target:
        print(i)
        break
else:
    print(-1)
```

```python
n = int(input())
factors = list()
for i in range(1, n+1):
    if n%i==0:
        factors.append(i)
if len(factors) == 2:
    print(f"{n} is prime")
else:
    print(f"{n} is not prime")
```

```python
#prime number
n = int(input())
for i in range(2,n//2+1):
    if n % i == 0:
        print(f"{n} not prime")
        break
else:
    print(f"{n} is prime")
```

- ### Nested Loop

```python
m = int(input())
for n in range(m):
    for i in range(2,n//2+1):
        if n % i == 0:
            print(f"{n} not prime")
            break
else:
    print(f"{n} is prime")
```
