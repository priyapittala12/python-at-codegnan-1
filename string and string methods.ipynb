## String Methods
-Case Converion Methods
    -upper(): It ccoverts entire string into upper case
    -lower(): It converts entire string intlo lower case
    -title(): It converts the first letter of every word into upper and remaining lower
    -swapcase(): It converts upper case intlo lower case and lower to upper
    -capitalize(): It converts first character into upper case and remaining intlo lower case
    

```python
string = "The sky is blue  and ocean is blue"
print(string.upper())
print(string.lower())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
- Remove wwhite space or pattern at ends of string 
    -strip():By default it removes the white space at both ends of string
    -istrip(): It removes the white space at left side of string
    -rstrip(): It removes the white space at right side of string
```

```python
s = "       codegnan       "
print(s)
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s = "@@@@@code@@gnan@@@@@"
print(s)
print(s.strip("@"))
print(s.lstrip("@"))
print(s.rstrip())
```

```python
# split(): It splits the entire strin into list of substing based on specified seperator
# by default it splits at space 
string = "The sky is blue and ocean is blue"
print(string.split())
print(string.split("is"))
print(string.split("e"))
```

```python
# 'sep'.join(iterable): It joins the all iterable elements with sperator
words = ['The', 'sky', 'is', 'blue', 'and', 'ocean', 'is', 'blue']
print("".join(words))
print(" ".join(words))
print("-".join(words))
```

```python
# Substring Index finding
# index(substring):If subtsring is exists in string, it returns the first occurence of substring  position, otherwise 'valueError'
# find(substring):If subtsring is exists in string, it returns the first occurence of substring  position, otherwise -1
```

```python
string = "The sky is blue  and ocean is blue"
print(string.index("is"))
print(string.index("z"))
```

```python
string = "The sky is blue  and ocean is blue"
print(string.find("is"))
print(string.find("z"))
print(string.find("is", 10))# it starts checking from 10th position
print(string.find("e", 10))
print(string.find("e", -5))
```

```python
# count(substring):It returns the total non-overlapping occucences of substring
s = "ABCBCBCbbaabb"
print(s.count("B"))
print(s.count("BCB"))
lower_string = s.lower()
print(lower_string)
print(lower_string.count('b'))
print(s.upper().count('B'))
print(s.count('z'))
```

```python
# replace(old_stbstring, new_substring): It replace the old sub_string with new sub string
string = "The sky is blue  and ocean is blue"
print(string.replace("blue", "Red"))
print(string.replace('e', 'E'))
print(string.replace('e' ,'E', 2))# it replace the first 2 occurance

```

```python
### String checking methods
 **String checking methods returns the Boolean value as result**
-isupper(): It returns True, If entire string is in upper case
-islower(): It returns True, If entire string is in lower case
-isalpha(): It returns True, If entire string contains alphabets
-isdigit(): It returns True, If entire string contains digits
-isalnum(): It returns True, If entire string contains either alphabets or digits
-startswith(substring): It returns True, If string starts with  substring
-endswith(substring): It returns True, If string ends with substring
```

```python
print("CODE".isupper())
print("CODE".islower())
print("code".islower())
print("1234".isdigit())
print("123abc".isdigit())
print("CODE".isalpha())
print("code123".isalnum())
print("1234".isalnum())
print("CODE".startswith('C'))
```
