```python
# check whether the product stock is available or not(read product name and stock)
product = input()
stock = int(input())
if stock > 0:
    print(f'currently {product} : {stock} items available')
else:
    print(f'currently {product} out of stock')
```

```python
# in a shop contains some items along with prices like{'pen':5, 'book':20, 'pencil':}
products = {'pen':5, 'book':20, 'pencil':3}
item = input().lower()
quantity = int(input())
if item in products:
    total_amount = products.get(item)*quantity
    print(f'total amount of {quantity} {item} is: {total_amount}')
else:
    print(f"{item} not available")
    
```

 - ### Nested If

```python
# check whether the number is even or odd
n = int(input())
if n>=0:
    if n%2==0:
        print(f'{n} is +ve  even')
    else:
        print(f'{n} is +ve odd')
else:
    if n<=0:
        if n%2==1:
            print(f'{n} is -ve even')
        else:
            print(f'{n} is -ve odd')
    
```

```python
n1, n2, n3 = map(int, input().split())
if n1>n2:
    if n1>n3:
        print(n1)
    else:
        print(n3)
else:
    if n2>n3:
        print(n2)
    else:
        print(n3)
        
    

```
