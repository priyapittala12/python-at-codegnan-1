```python
# check whether number the number is divisible by both 3 and 5
num = int(input())
if num%5==0 or num%3==0:
    print(f'{num}is divisible by either  3 & 5')
else:
    print(f'{num}is not  divisible by either 3 & 5')
```

```python
#generate total bill if total bill is more than 500 then give 10% discount otherwise 0% discount
total_bill = int(input())
if total_bill>500:
    discount = total_bill * 0.1
    print(f'Tota bill after 10% discount: {total_bill - discount}')
else:
    print(f'total bill after0% discount: {total_bill}')
    
```

```python
# check whether the product stock is available or not(read product name and stock)
product = input()
stock = int(input())
if stock > 0:
    print(f'currently {product} : {stock} items available')
else:
    print(f'currently {product} out of stock')
```

```python
# in a shop contains some items along with prices like{'pen':5, 'book':20, 'pencil':}
products = {'pen':5, 'book':20, 'pencil':3}
item = input().lower()
quantity = int(input())
if item in products:
    total_amount = products.get(item)*quantity
    print(f'total amount of {quantity} {item} is: {total_amount}')
else:
    print(f"{item} not available")
    
```
