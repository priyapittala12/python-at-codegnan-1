- ### REGULAR EXPRESSIONS

```python
import re 
```

```python
# match 
string = "The sky is blue and ocean is blue"
res = re.match("The", string)
print(res)
print(res.span()) #it returns the matched pattern position
print(res.group()) # it returns the matched result
```

```python
# match(pat, string): it checks whether the string starts with pattern or not
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("sky", string)
    print(res)
    print(res.span()) #it returns the matched pattern position
    print(res.group()) # it returns the matched result
except Exception as e:
    print("There is no matched pattern")
```

```python
# search(pat, string): it checks the entire string and it returns the first matched pattern 
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("sky", string)
    print(res)
    print(res.span()) #it returns the matched pattern position
    print(res.group()) # it returns the matched result
except Exception as e:
    print("There is no matched pattern")
```

```python
# search
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("z", string)
    print(res)
    print(res.span()) #it returns the matched pattern position
    print(res.group()) # it returns the matched result
except Exception as e:
    print("There is no matched pattern")
```

```python
# findall(pat, string): it returns all matched pattrens as list. 
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("blue", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[bcdea]+", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# split(pat, string): It splits the string wherever the pattern is matched
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("blue", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# split
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("[abc]", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[isabce]+", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall
try:
    string = "The sky is blue and ocean is bl9ue"
    res = re.findall("[a-zA-Z]+", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall
try:
    string = "The sky is blue 12 and ocean is bl9ue 2376"
    res = re.findall("[0-9]+", string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
# validate mobile number
try:
    mobile = input()
    res = re.fullmatch("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
except:
    print("There is no matched pattern")
```

```python
# validate mobile number
try:
    pan = input()
    res = re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]$", pan)
    print(res)
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is invalid")
except:
    print("There is no matched pattern")
```

```python
# validate aadhar number
try:
    aadhar = input()
    res = re.fullmatch(r"^[2-9][0-9]{3}\s?[0-9]{4}\s?[0-9]{4}$", aadhar)
    print(res)
    if res:
        print(f"{aadhar} is valid")
    else:
        print(f"{aadhar} is invalid")
except:
    print("There is no matched pattern")
```
