- ### LIST COMPREHENSION

```python
# generate 1-n numbers using list comprehension
n = int(input())
res = [x for x in range(1, n+1)]
print(res)
```

```python
# first n even numbers using list comprehension
n = int(input())
res = [x*2 for x in range(0, n)]
print(res)
```

```python
# find the first n qubic value
n = int(input())
res = [x**3 for x in range(1, n+1)]
print(res)
```

```python
# find the odd numbers between n numbers
n1,n2 = map(int, input().split())
res = [x for x in range(n1, n2+1) if x%2 == 1]
print(res)
```

```python
# in given arr of integers perform following operations
# if even - find square of a number
# if odd - find squareroot value(upto 2 decimal places)
arr = list(map(int, input().split()))
res = [x**2 if x%2==0 else round(x**0.5, 2) for x in arr]
print(res)
```

```python
# 2 dice combination values
res = [ (x,y) for x in range(1,7) for y in range(1,7)]
print(res)
```

```python
# return the dice combination if sum of two face value is divisible by 3
res = [(x,y) for x in range(1,7) for y in range(1,7) if (x+y)%3==0]
print(res)
```

```python
res = [(x,y) for x in range(1,7) for y in range(1,7) if x == 4 or y == 4]
print(res)
```

```python
# count the number of vowels exists in string
string = input()
res = [char for char in string.lower() if char in"aeiou"]
print(res)
print(len(res))
```

```python
string = input()
res = sum([1 for char in string.lower() if char in"aeiou"])
print(res)
```

- ### SET COMPREHENSION

```python
# set comprehension
# 1 to n numbers
n = int(input())
res = {x for x in range(1, n+1)}
print(type(res))
print(res)
```

```python
# dict comprehension
# 1 to n numbers
n = int(input())
res = {x:x**2 for x in range(1, n+1)}
print(type(res))
print(res)
```

```python
# Tuple comprehension 
# 1 to n numbers
n = int(input())
res = tuple(x for x in range(1, n+1))
print(type(res))
print(res)
```

```python
# weather the person is eligible for vote or note
age = int(input())
if age >= 18:
    print("Eligible")
else:
    print("Not Eligiblr")
```

```python
# assert keyword : This keyword is used for testing
age = int(input())
assert age >= 18, "Age must be above 17"
print("Eligible")
```

```python
# reverse a string without using slicing
string = input()
rev = ''
for char in string:
    rev = char + rev
print(rev)
```

```python
n = int(input())
res = ""
while n>0:
    rem = n%2
    res = str(rem) + res
    n = n // 2
print(res)
rev = ""
for char in res:
    rev = char + rev
print(rev)
print("palindrome" if res == rev else "Not Plindrome")
```
