```python
# Define list
# List is ordered collection 
lst = [1,2,4.5,"Hi","Hello",1,3,4]
print(lst)
print(lst[4])
print(lst[1:6:1])
print(lst[-5])
```

```python
# slicing
lst = [10, 20, 30, 40, 50, 60, 70]
print(lst[:5:-1])
print(lst[:500])
print(lst[::-1]) #Reverse list
```

```python
# List operations
# 1. List concatination (using + operator)
# 2. List repetation (using * operator)
l1 = [1,2,3,]
l2 = [3,4,5,6]
print(l1+l2)
print(l1+l1)
print(l1 * 3)
print([0] * 10)
```

```python
### List methods
-Item adding Method
    -append(item) : It adds item at the end of the list
    -extend(iterable) : It adds the all iterator element at end of list
    -insert(pos,item) : It adds item at specified position(it shifts one position right)
```

```python
lst = [10,20,30,40]
lst.append(100)
print(lst)
lst.extend([1,2,3])
print(lst)
lst.append([100,200])
print(lst)
lst.insert(0,1000)
print(lst)
```

```python
### Item Reoving Methods
    -remove(item): It removes the first occurance item from the list if it is exists in list otherwise 'valueError'
    -pop(pos) : It remove and return the specified position item if position is passed, otherwise it remove and return the last item from list 
    -clear(): It removes all elements from the list 
```

```python
lst = [10,20,30,20,40]
print(lst.remove(20))
lst.remove(100)
```

```python
lst = [10, 20, 30, 20, 40, 50, 20]
val = lst.pop()
print(val)
print(lst)
val2 = lst.pop(0) # it removes the 0th position
print(val2)
print(lst)
lst.clear()
print(lst)
```

```python
# index and count()
    # index(item): If item is present in list it returns the Index otherwise valueError
    # count(item) : It return the count of Total occurance of item
lst = [10, 20, 30, 20, 40, 50]
ind = lst.index(20)
print(ind)
print(lst.count(20))
```

```python
# sorting items
# sort(reverse = False) : By default it sorts the origina list in into Ascending order. If reverse is True it sorts in Decending order
lst = [1,4,2,5,-5,3,-45,-4,30]
lst.sort()
print(lst)
lst.sort(reverse = True)
print(lst)
```

```python
# sorting items
# sort(reverse = False) : By default it sorts the origina list in into Ascending order. If reverse is True it sorts in Decending order
lst = ['hi', 'hello', 'banti']
lst.sort()
print(lst)
lst.sort(reverse = True)
print(lst)
```
