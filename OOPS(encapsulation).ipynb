    `syntax:`  <br>
           `    class class_name:`<br>
           `        @classmethod`<br>
           `       def method_name(cls)`:<br>

-  static method: The method defination starts with `staticmethod` decorator and it dosen't take the first parameter as `self or class` those methods called as `Static method`. This method works on except instance attributes and class attributes.
     
     `syntax:`<br>
           `    class class_name:`<br>
           `        @staticmethod`<br>
           `       def method_name(parameters)`:<br>
           `           # block code`<br>

- #### Encapsulation:
      - Encapsulation means, wraping the methods and attributes in single unit that one is class. The main aim of Encapsulation is to restrict the direct access of attributes and methods.
        - This type of Restriction can implement by using `Access Modifier`
        - Types of Access modifiers
            1. public
            2. protected     # starts with single underscore
            3. private       # this starts with dunderscore
      - In python we don't have any keywords to represent it public, protected and private, but by using underscore(_) can define.
            1. public: The public members(i.e attribute + methods) dosen't starts with a underscore. This public method can access any where in the program.
            2. protected: The protected members(i.e attribute + methods) starts with single underscore. This member can access in inside the class and in Subclasses also.
            3. private: The private members(i.e attribute + methods) starts with double underscore(__). This member can access only inside the class
- ### Accessing levels of Access modifiers:
        - public: inside class,sub class, outside the class.
        - protected: inside class, sub class.
        - private: only inside the class.

```python
class Bank:
    def __init__(self, name, account, password, balance):
        self.name = name
        self.account = account
        self.password = password
        self._balance = balance
    def __total_users(self):  #getter method
        return "Total users : 111"
    def get_balance(self): # getter method
        return f"Current Balance is:{self._balance}"
    def deposite(self, deposite_amount):  # setter method
        self._balance += deposite_amount
        return f" {deposite_amount} Deposited Successfully"
    
        
u1 = Bank("Priya", 1234, "priya@135", 20000)
print(u1.name)
print(u1.get_balance())
print(u1.deposite(30000))
print(u1.get_balance())
print(u1._balance)  # protected members access in outside the class
print(u1._Bank__total_users()) #Name manling: syntax: obj._classname__attribute
```
