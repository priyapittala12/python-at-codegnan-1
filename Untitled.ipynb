```python
# indexing
s = "codegnan"
print(s[2])
print(s[-3])
```

```python
# slicing
# str_obj[start:end:step]
s = "codegnan"
print(s[2:7])
print(s[-4:])
print(s[5:2])
print(s[3:100])
print(s[::-1])
print(s[-3::-1])
print(s[-3::1])
```

```python
# empty string Representation
# type() : This function returns the objects datatype
s1 = ""
s2 = str()
print(type(s1))
print(type(s2))
```

```python
# string reading from user
# input() :This function is used to read dynamically from user and this function by default read it as string datatype
s = input()
print(type(s))
print(s)
```

```python
# String operations
# 1.String Concatenation (using + operator)
# 2. String Repetation (using * operator)
s1 = "code"
s2 = "Gnan"
s3 = s1 + s2
print(s3)
print(s1+s1+s1)
print(s1 * 100)
print((s1+s2)*60)
```

```python
# immutable
s = "codegnan"
res = s[:1] + 'z' + s[2:]
print(res)
```
