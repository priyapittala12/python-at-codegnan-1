- `POLYMORPHISM`:
     - In python `"Polymorphism"` `'Poly'` means `'many'` amd `'morphism'` means `'forms'`
     - In polymorphism single function  shows its different behavioural based on input context.
     - In python polymorphsim can achieve through the following ways
         1. Polymorphsim through operators
         2. Polymorphsim through functions
         3. Polymorphsim through methods  <br>
     `Polymorphsim through has 2 methods:` <br>
       ` i. Runtime polymorphsim `- method overriding  <br>
        `ii. Compile time polymorphsim` - method overloading <br>

 `1. POLYMORPHSIM THROUGH OPERATORS:`
        - Here single operator shows different actions based on input context/values.
        
           `ex:`  
                - '+' operator performs addition operation when input are integers, as well as it forms string concatination when inputs are strings. 
             `ex:` - 3+5 = 8  --addition 
                   - "Code" + "Gnan"  `--string concatination` 
                   - 3*5 = 15 `--multiplication operation`
                   - 3*code `--repetation ` 
`2. POLYMORPHSIM THROUGH FUNCTIONS:`
        - Here a function shows different actions based on input context 
    `ex:` <br>
        - len('Hello') `--string length`  <br>
        - len([1,2,3,4,5]) `--list length` <br>

`3. POLYMORPHSIM THROUGH METHODS:` <br>
              - `i. METHOD OVERRIDING`: ( through the inheritance) <br>
                         - A method defined in child class that is already exists in parent class with the same name.<br>
                         - Whenever if you call that method through child class method, the child class method overrides the parent class method.<br>
                         - This is known as method `OVERRIDING`. <br>

                         `syntax`: <br>

                                 class A
                                     def show:
                                 class B(A):
                                     def show()
                                 b = B()
                                 b.show()

              -` ii. METHOD  OVERLOADING`: 
                        - A class contains multiple methods with same name, eith different parameters list, In that case last defined method overrides the previous defined method. This is called as `OVERLOADING`
                        - Directly python dosen't supports method overloading but we can achiev same functionality through default arguments, variables, length positional, or variable length keyword arguments.
                                
    



```python
# Method Overriding:
# Even if you define different parameters in methods, it always child class method overridies parent class method.
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self, location): # child class
        return f"My name is {self.name} and i am staying in{location}"

class Student(Person):
    def __init__(self, name, age, gender, branch, rollno):
        super().__init__(name, age, gender)
        self.branch = branch
        self.rollno = rollno
    def details(self):
        print(super().details())
        return f"My name is {self.name} and i am belongs to {self.branch} branch"
    
s = Student("Priya", 21, "Female", "ECE", 101)
print(s.details()) 
```

```python
# Method Overloading
# Directly python dosen't supports method overloading but, we can achieve same functional through the default arguments or Arbitrary Arguments. 
class Math:
    
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# Magic Methods or Special Methods
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # string magic method
        return f"This is Point class and Point value is {self.x}" 
    def __add__(self, other):
        return (self.x + other.x)
    def __sub__(self, other):
        return (self.x - other.x)
    def __mul__(self, other):
        return (self.x * other.x)
    def __gt__(self, other):
        return (self.x > other.x)
    def __lt__(self, other):
        return (self.x < other.x)
    def __or__(self, other):
        return (self.x|other.x)
    def __mod__(self, other):
        return (self.x % other.x)

        
p1 = Point(5)
p2 = Point(3)
print(p1+p2)
print(p1-p2)
print(p1*p2)
print(p1>p2)
print(p1<p2)
print(p1|p2)
print(p1%p2)

```
