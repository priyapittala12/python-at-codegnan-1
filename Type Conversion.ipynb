```python
# integer to float, boolean, string
num = 10
print(float(num))
print(bool(num))
print(str(num))
print(complex(num))
```

```python
# float to interger, boolean, string
val = 10.2556
num = int(val)
print(type(num), num)
print(bool(val))
str_val = str(val)
print(type(str_val), str_val)
print(str_val[:4])
print(str_val[1], str_val[-1])

```

```python
# boolean to integer, float, string
b = True
print(int(b))
print(float(b))
print(str(b))
print(str(b)[1])
```

```python
# string to int, float, bool, list, tuple, set
# if string value is number
s = '123456'
print(int(s))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

```python
# string to int, float, bool, list, tuple, set
# if string value is float
s = '123456.1205'
print(int(float(s)))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

```python
# list to string, tuple, set, dictionary
s = [12.1, 12, 34, 56, 67, 78]
print(str(s))
print(tuple(s))
print(set(s))
l = [('a',1), ('b',2)]
print(dict(l))
```

```python
# tuple to string, list, set
t = (1,2,3,'priya',4)
print(str(t))
print(list(t))
print(set(t))
```

```python
# set to string, list, tuple
s = {23,34,56,'hi',33}
print(str(s))
print(list(s))
print(tuple(s))
```
