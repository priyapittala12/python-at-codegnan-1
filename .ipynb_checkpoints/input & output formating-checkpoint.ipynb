- ### Input Formatting

```python
# reading intege r from user
num = int(input("Enter a number:"))
print(type(num))
print(num)
```

```python
# reading float value from user
val = float(input())
print(type(val))
print(val)
```

```python
# complex reading from user
val = complex(int(input()), int(input()))
print(type(val))
print(val)
```

```python
# boolean reading
val = bool(input())
print(type(val))
print(val)
```

```python
# boolean reading as string
val = bool(int(input()))
print(type(val))
print(val)
```

```python
# reading string value from user
val = str(int(input()))
print(type(val))
print(val)
```

```python
# reading list value from user
val = list(str(input()))
print(type(val))
print(val)
```

```python
# reading tuple value from user
t = tuple(set(input()))
print(type(val))
print(val)
```

```python
# reading list of integers from user
lst = list(map(int, input().split()))
print(lst)
```

```python
# reading list of strings from user
lst = list(map(str, input().split()))
print(lst)
```

```python
#approch 2
lst = input().split()
print(lst)
```

```python
# reading list of tuple from user
lst = list(map(tuple, input().split()))
print(lst)
```

```python
# reading tuple of integers from user
lst = tuple(map(int, input().split()))
print(lst)
```

```python
# eval(): This functions evaluates the string type expression.
val = eval(input())
print(type(val))
print(val)
```

```python
print("3+5")
print(eval("3+5"))
```

- ### Output Formating

```python
# string using concatination
name = "priya"
age = 21
print("my name is" + name + "and i am " +str ( age )+ "years old")
```

```python
# string formatting using comma
name = "priya"
age = 21
print("my name is",name, "and i am ", age, "years old")
```

```python
# Access specifiers
# string - %s
# integer - %d
# float - %f
name = "priya"
age = 21
print("my name is %s and i am %d years old"%(name, age))
```

```python
name = "priya"
age = 21
print("my name is %s and i am %d years old"%(age, name))
```

```python
# 3 dot format
name = "priya"
age = 21
print("my name is {} and i am {} years old". format(name, age))
print("my name is {} and i am {} years old". format(age, name))
```

```python
# 3 f-format
name = "priya"
age = 21
print(f"my name is {name} and i am {age} years old")
print(f"my name is {age} and i am {name} years old")
```
