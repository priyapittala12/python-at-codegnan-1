```python
# characteistics
# ordered collection
t = (1,2,3,[10,20,30],23)
print(t)
```

```python
# immutable
# After operation it dosen't change the length of objectt and dosen't change remaining element position
t = (1,2,3,[10,20,30],23)
t[3].append(100)
print(t)
```

```python
# Empty tuple representation
# single element representation
t1 = ()
t2 = tuple()
print(type(t1), type(t2))
```

```python
# single element representation
t1 = (10)
t2 = (10,)# if we specify ',' then it is tuple
l1 = [10]
print(type(t1), type(t2), type(l1))
```

```python
(0,) * 10
```

```python
# Tuple operations
# Tuple concatination
# Tuple Repetation
t1 = (1,2,3,45,[10,20,30],12)
t2 = (2,4,"hi")
t3 = (t1 + t2)
print(t3)
t1 = (34,56,7,[1,2,3,4],'priya')
t2 = (2,3,4,[100,23,45],'hello')
t3 = (t1 * 3)
print(t3)
```

```python
# Tuple Method
# index
# count()
t1 = (100,200,'hello',400)
ind = t1.index(200)
print(ind)
print(t1.count(200))
```

```python
# Buitin Functions- len(), min(), max(), sum()
t1 = [1,2,3,4,5]
print(len(t1),min(t1),max(t1),sum(t1))
print("Average:",sum(t1) / len(t1))
```
