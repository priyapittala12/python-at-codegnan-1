- ### Built In Function

```python
# round(val, pos):
print(round(10.25684, 2))
print(round(10.25499, 2))
print(round(10.25555, 2))
```

```python
# chr() : This function returns the character based on ASCII value
# ord() : It returns the ASCII value based on character
print(chr(65))
print(ord('A'))
print(chr(97))
print(ord('1'))
print(chr(35))
print(ord('4'))
```

```python
# abs() :
print(abs(10-20))
```

```python
s1 = "CODEA"
s2 = "GNANA"
n = len(s1)
s = ""
for i in range(n):
    diff = abs(ord(s1[i]) - ord(s2[i]))
    if diff == 0:
        s += s1[i]
    else:
        s += chr(64+diff)
print(s)
 
```

```python
# find bigger and smaller character from s1 and s2
# compare each character in both string and if s1[i] > add s1[i] to bigger and s2[i] to smaller
# if both are same than leave it
s1, s2 = map(str, input().upper().split())
n = len(s1)
big = small = ""
for i in range(n):
    if s1[i] != s2[i]:
        if s1[i] > s2[i]:
            big += s1[i]
            small += s2[i]
        else:
            big += s2[i]
            small += s1[i]
print(big)
print(small)
    
```

```python
# map(fun, iterable): it applies the functionality on each element in iterable
#filter(fun, iterable): it filters the iterables, it returns the value which is not None 
# find the square of list of integers
def square(x):
    return x**2
lst = [1,2,3,4,5]
print(list(map(square, lst)))
```

```python
# filter the even numbers from list of integers
def even(x):
    if x%2 == 0:
        return x
        
arr = [1,2,3,4,5,6,7]
print(list(filter(even, arr)))
```

```python
def square(t):
    x, y = t
    return x + y
lst = [(1,2),(3,4)]
print(list(map(square, lst)))
```

```python
# zip(iterable1, iterable2, ....., iterableN):
# it returns list of tuples, the tuple is generated as each position elements from all
l1 = ['a','b','c']
l2 = [1,2,3,4,5]
print(list(zip(l1, l2)))
```

```python
# enumerate(iterable): it returns the list of tuple of index and value
arr = [10,20,30]
print(list(enumerate(arr)))
```

```python
# find the target element position in list
arr = [10, 20, 30, 40, 50]
target = 40
for ind, val in enumerate(arr):
    if target  == val:
        print(ind)
        break
else:
    print(-1)
```

```python
# any(iterable): it returns True if any one iterable element in the iterable boolean value isTrue
arr = [1,2,3,0,0,0]
print(any(arr))
print(any([0,0,0,0]))
# all(iterable): it returns True if all iterable element in the iterable boolean value isTrue
print(all([1,2,3,4]))
print(all([True, False, True]))
print(all([False, False, False]))
```

```python
# isinstance(obj, class_name): It checks weather the object is belongs to particular class or not
print(isinstance(10, int))
print(isinstance(12.25, int))
```

```python
# sorted(): it return the new sorted object, by default it returns the descending order
# reverse(): it returns the reverse of an object value
arr = [10,2,34,35,3,5,6]
res = sorted(arr, reverse = True)
print(res)
print(arr)
print(list(reversed(arr)))
print("".join(list(reversed("Hello"))))
```

```python
arr = ["Hello", "Hi", "Apple", "Banana"]
sorted(arr, key = len)
```

```python
# count the frequency of characters in string 
string = "pittala"
for char in string:
    print(f"{char} - {string.count(char)}")
```
