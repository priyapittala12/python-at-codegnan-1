```python
n = 5
for i in range(n):
    for j in range(n):
        if i==j or j==n-i-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# inverse right angle triangle
n = 5
for i in range(n):
    for j in range(n):
        if i<=j:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# right angle triangle
n = 5
for i in range(n):
    for j in range(n):
        if i>=j:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# reverse of right angle triangle
n = 5
for i in range(n):
    for j in range(n):
        if j>=n-i-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# inverse reverse right angle triangle
n = 5
for i in range(n):
    for j in range(n):
        if j<=n-i-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```
