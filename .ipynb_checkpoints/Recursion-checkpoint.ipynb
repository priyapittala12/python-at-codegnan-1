```python
# print 1 to n using recursion
def func(x):
    if x == 0:
        return 0
    func(x - 1)
    print( x)
print(func(5))
```

```python
# print 1 to n using recursion
def func(x):
    if x == 0:
        return 0
    print(x)
    func(x - 1)
func(5)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```

```python
# sum of digits in a number
# n = 456 o/p: 15
def sumofdigits(n):
    if n == 0:
        return 0
    return n % 10 + sumofdigits(n//10)
print(sumofdigits(456))
print(sumofdigits(15346))
        
```

```python
# sum of digits in a number
# n = 456 o/p: 15
def sumofdigits(n):
    if n == 0:
        return 0
    return  1 + sumofdigits(n//10)
print(sumofdigits(456))
print(sumofdigits(15346))
        
```

- ### Indirect Recursion

```python
# find the square of a number using lambda
s = lambda x:x**2
print(s(10))
```

```python
# find the square of list of integers using map and lambda
arr = [1,2,3,4,5]
print(list(map(lambda x: x** 2, arr)))
```

```python
# find the even numbers
arr = [1,2,3,4,5]
print(list(filter(lambda x: x%2==0, arr)))
```

```python
# find weather the number is even or odd
s = lambda x: "even" if x%2==0 else "odd"
print(s(10))
```

```python
# from list of numbers check weather the number is even or odd
arr = [1,2,3,4,5]
# 1 - odd
# 2 - even
print(list(map(lambda x:f"{x} - even" if x%2==0 else f"{x} - odd", arr)))
```

```python
string = input()
count = {}
for char in string:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)
```

```python
string = input()
count = {}
for char in string:
    if char in count:
        count[char] = count.get(char, 0) + 1
    else:
        count[char] = 1
print(count)
```
