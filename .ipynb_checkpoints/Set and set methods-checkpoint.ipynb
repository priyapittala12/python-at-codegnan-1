- ### Set Chracteristics

```python
# set unordered collection
# set returns unique elements
s = {1,2,3,2,3,'hi', 'a', 2.02, (10,20,30)}
print(s)
# print(s[0]) # set does not supports indexing
```

```python
# list can't be used as a set element
s = {1,2,[3,2,3],'hi', 'a', 2.02, (10,20,30)}
print(s)
```

- ### Set Methods

```python
# Elements adding method
# add(ele) :It adds the element to the set if element is not in set, otherwise nothing
s = {1,2,3,4,5,4,3}
print(s)
s.add(100)
s.add(5)
s.add("code")
print(s)
```

- ### Element removing methods
- ### Remove(item): It removes the item if it exists in set otherwise 'KeyError'
- ### discard(item): It removes the item if it exists in set otherwise nothing


```python
s = {1,2,3,4,5}
s.remove(5)
print(s)
# s.remove(100) # KeyError
s.discard(3)
print(s)
s.discard(100) 
print(s)
```

```python
# Sets Operations
# union(set): It returns all the unique elements from all the sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
union = s1.union(s2,s3)
print(union)
res = s1 | s2 | s3
print(res)
```

```python
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
intersection = s1.intersection(s2,s3)
print(intersection)
intersection = s1.intersection(s2)
print(intersection)
res = s1&s2&s3
print(res)
```

```python
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
difference = s1.difference(s2,s3)
print(difference)
difference = s2.difference(s1)
print(difference)
res = s1-s2-s3
print(res)
print(s1|s2 -s3)
```

```python
# set operstions
#symmetric_difference(set): It return the non common elements from all the set
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
sy_diff = s1.symmetric_difference(s2)
print(sy_diff)
res = s1^s2^s3
print(res)
```

```python
# issubset(set)
#issuperset(set)
# isdisjoint(set)
s1 = {1,3,5,7}
s2 = {1,3,5,7}
s3 = {5,1,2,8,9}
issubset = s1.issubset(s2)
print(issubset)
issubset = s3.issubset(s2)
print(issubset)
issuperset = s1.issuperset(s2)
print(issuperset)
issuperset = s2.issuperset(s3)
print(issuperset)
isdisjoint = s1.isdisjoint(s2)
print(isdisjoint)
isdisjoint = s2.isdisjoint(s3)
print(isdisjoint)
s = s1,s2,s3
print(dir(s))



```
