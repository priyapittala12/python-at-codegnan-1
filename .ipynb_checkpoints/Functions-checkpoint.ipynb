- ### Functions

```python
#Types on functions based on parameters and return value
# 4. with parameters and with return

# function defination block
def addition (x:int, y:int)->int:
    c = x + y
    return c
# function calling block
a, b = 10,20
res = addition(a, b)
print(res)
```

```python
# 3. with parameters and without return
# by default function returns None value, when function not returning any value
# function defination block
def addition (x:int, y:int)->int:
    c = x + y
    print(f"This is from Function defination block:{c}")
# function calling block
a, b = 10,20
res = addition(a, b)
print(f"This is from Function defination block:{res}")
```

```python
# 2. without parameters and with return.
# function defination block
def addition ()->int:
    x = 10
    y = 20
    z = x+y
    return z
# function calling block
res = addition()
print(f"This is from Function defination block:{res}")
```

```python
# 1. without parameters and without return
# function defination block
def addition ()->None:
    x = 5
    y = 10
    c = x + y
    print(f"This is from Function defination block:{c}")
# function calling block
res = addition()
print(f"This is from Function defination block:{res}")
```

```python
# implement even or odd using function
def evenodd(x:int):
    if x % 2 == 0:
        return "Even"
    else:
        return "Odd"
a = 10
res = evenodd(a)
print(res)
        
```

```python
# implement prime numbers using functions
def check_prime(n:int)->str:
    if n < 2:
        return "Not Prime"
    for i in range(2, n//2+ 1):
        if n % i == 0:
            return "Not Prime"
    return "Prime"
# function calling block
num = int(input())
print(check_prime(num))
```
