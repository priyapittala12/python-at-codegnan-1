- ## Dictionary Characteristics

```python
# Dictionar ordered collection
details ={"Name": "Priya", "Course":"Python", "Batch":57, "Age":21}
print(details)
```

```python
# Dictionary values access by using keys
# updating dictionary values
# adding new items to dictionary
details ={"Name": "Priya", "Course":"Python", "Batch":57, "Age":21}
print(details["Course"])
#peint (details['location']) #keyError: 'Location'
#updating dictionary values
details['age'] = 21
print(details['age'])
#adding new items
details['Location'] = "Hyderabad"
print(details)
```

```python
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21}
print(details)
```

```python
#list can't be used as dictionary key
#dictionary keys are unique
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21, "Marks":[10,20,30]}
print(details["Marks"])
```

- ## Dictionary Methods
    - get(key, default_valur:optional):it returns the value if key is exists in dictionary otherwise it returns default value

```python
#Dictionary keys are unique
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21, "Location":"Hyderabad"}
print(details["Name"])
# print(details["Marks"])#this one raises key error
print(details.get("Name"))
print(details.get("Marks"))# get method bydefault it returns none if key not exists in 
print(details.get("Marks", "Mrks not available"))#it returns default value because,
```

```python
- update(new_dictionary):n This method updates the value if new_dictionary keys already exists in dictionary,otherwise it adds new items to dictionary
```

```python
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57}
details.update({'batch':58, 'course':'java', "Location":"Hyderabd"})
print(details)
```

- keys(): This method returns the list dictionary keys
- values():This method returns the list dictionary values
- items(): This method returns the List of Tuple of key-value pair

```python
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21, "Location":"Hyderabad"}
keys = details.keys()
print(details.keys())
print(type(keys))
print(keys)
print(details.values())
print(details.items())
```

- popitem(): It returns and removes the last item as tuple
- pop(key): This method removes the specified key item and return the specified key's value, otherwise KeyError

```python
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21, "Location":"Hyderabad"}
item = details.popitem()
print(item)
print(details)
val = details.pop("Name")
print(val)
print(details)
```

```python
# Membership
# it checks with dictionary keys
details ={"Age":21, "Name": "Priya", "Course":"Python", "Batch":57, "Age":21, "Location":"Hyderabad"}
print("Priya" in details)
print("Name" in details)
```
