- ### INHERITANCE

- In python, A class get properties from another class this processes called as "Inheritance"
- A child class is inherited from parent class.
     - Types of inheritance:
         1. Single inheritance
         2. Mmulti level inheritance
         3. Multiple inheritance
         4. Hierarchical inheritance
         5. Hybri inheritance

1. Single inheritance:
           - A child class get properties from single parent class. This is called as "Single inheritance"
     `syntax: <br>
              class parent_class_name:  <br>
                  # attributes  <br>
                  # methods  <br>
              class childclass_name(parent_class_name):  <br>
                  # attributes  <br>
                  # methods   <br>

2. Multilevel inheritance:
             - A class(i.e child class) get properties from another class(i.e class) that get properties from same another clss(i.e grandparent)

3. Multiple inheritance:
             - A single child class get properties from multiple parent classes. This is called as "Multiple inheritance"  

4. Hierarchical inheritance:
             -Multiple child classes get properties from single parent class.

5. Hybrid inheritance:
             - This is a combination of two or more types of inheritance.

```python
# Types of inheritances
# Single inheritance
# without child class implementation
# When child class donot have constuctor, in that case, parent class constructor behave like child class constuctor.
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and i am {self.age} years old"

class Student(Person):
    pass
s = Student("Priya", 21, "Female")
print(s.name)
print(s.details()) 
    
```

```python
# Single inheritance
# child class with constructor
# When child class donot have constuctor, in that case, parent class constructor behave like child class constuctor.
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and i am {self.age} years old "

class Student(Person):
    def __init__(self, stdid, name, age, branch):
        self.name = name
        self.age = age
        self.stdid = stdid
        #self.gender = gender
        self.branch = branch
    
s = Student(101, "Priya", 21, "ECE")
print(s.name)
print(s.branch)
print(s.age)
print(s.stdid)
print(s.gender) # AttributeError: 'Student' object has no attribute 'gender'
print(s.details())# It raises AttributeError because student Student reference dosen't contain gender attribute
    
```

```python
# Single inheritance
# child class with constructor
# When child class donot have constuctor, in that case, parent class constructor behave like child class constuctor.
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and i am {self.age} years old "

class Student(Person):
    def __init__(self, stdid, name, gender, age, branch):
        super().__init__(name, age, gender)
        self.stdid = stdid
        self.branch = branch
    
s = Student(101, "Priya", "Female",  21, "ECE")
print(s.name)
print(s.branch)
print(s.age)
print(s.stdid)
#print(s.gender) # AttributeError: 'Student' object has no attribute 'gender'
print(s.details()) 
    
```

```python
# 2. Multilevel innheritance
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is{self.name} and i am {self.age} years old"
    

class Student(Person):
    def __init__(self, name, age, branch,):
        super().__init__(name, age,)
        self.branch = branch
    

class Employee(Student):
    def __init__(self, name, age, branch, salary):
        super().__init__(name, age, branch)
        self.salary = salary
    def get_salary(self, bonous=0):
        self.salary += bonous
        return self.salary
        
e = Employee("Priya", 21, "HYD", 50000)
print(e.name)
print(e.branch)
print(e.details())
print(e.get_salary())
print(e.get_salary(bonous = 10000))
```
